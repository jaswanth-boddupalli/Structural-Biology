```python
import os
import pandas as pd
from rdkit import Chem
from rdkit.Chem import AllChem
from openbabel import openbabel as ob

# Function to convert SMILES to 3D structure and save as PDBQT
def process_smiles(compound_name, smiles, output_folder):
    try:
        # Convert SMILES to RDKit molecule
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            raise ValueError(f"Invalid SMILES string for {compound_name}")

        # Add hydrogen atoms
        mol = Chem.AddHs(mol)

        # Generate 3D coordinates
        AllChem.EmbedMolecule(mol, AllChem.ETKDG())

        # Optimize the 3D structure
        AllChem.UFFOptimizeMolecule(mol)

        # Compute Gasteiger partial charges
        AllChem.ComputeGasteigerCharges(mol)

        # Define file paths
        pdb_file = os.path.join(output_folder, f"{compound_name}.pdb")
        pdbqt_file = os.path.join(output_folder, f"{compound_name}.pdbqt")

        # Write the molecule to a PDB file
        with open(pdb_file, 'w') as f:
            f.write(Chem.MolToPDBBlock(mol))

        # Initialize Open Babel conversion
        ob_conversion = ob.OBConversion()
        ob_conversion.SetInAndOutFormats("pdb", "pdbqt")

        # Read the PDB file into an Open Babel molecule
        ob_mol = ob.OBMol()
        ob_conversion.ReadFile(ob_mol, pdb_file)

        # Write the molecule to a PDBQT file
        ob_conversion.WriteFile(ob_mol, pdbqt_file)

        # Remove the intermediate PDB file
        os.remove(pdb_file)

        print(f"Processed {compound_name} successfully.")

    except Exception as e:
        print(f"Error processing {compound_name}: {e}")

# Main function
def main():
    # Define input CSV file and output folder
    input_csv = 'Ligand_list.csv'
    output_folder = 'Ligand_3D_Structures'

    # Create output folder if it doesn't exist
    os.makedirs(output_folder, exist_ok=True)

    # Read the CSV file
    df = pd.read_csv(input_csv)

    # Process each row in the CSV
    for index, row in df.iterrows():
        compound_name = row['Compound Name']
        smiles = row['SMILES']
        process_smiles(compound_name, smiles, output_folder)

if __name__ == "__main__":
    main()
```

```python
import os
import csv
from rdkit import Chem
from rdkit.Chem import AllChem
import openbabel.pybel as pybel

# Define input CSV file and output directory
input_csv = 'Ligand_list.csv'
output_dir = 'Ligand_3D_Structures'

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Function to sanitize filenames by replacing problematic characters
def sanitize_filename(name):
    return "".join(c if c.isalnum() or c in (' ', '.', '_') else "_" for c in name)

# Read the CSV file
with open(input_csv, newline='', encoding='utf-8') as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        compound_name = row['Compound Name']
        smiles = row['SMILES']
        sanitized_name = sanitize_filename(compound_name)
        pdb_file = os.path.join(output_dir, f"{sanitized_name}.pdb")
        pdbqt_file = os.path.join(output_dir, f"{sanitized_name}.pdbqt")

        try:
            # Convert SMILES to RDKit molecule object
            mol = Chem.MolFromSmiles(smiles)
            if mol is None:
                print(f"Error processing {compound_name}: Invalid SMILES string.")
                continue

            # Add hydrogen atoms to the molecule
            mol = Chem.AddHs(mol)

            # Generate 3D coordinates
            AllChem.EmbedMolecule(mol, AllChem.ETKDG())

            # Optimize the 3D structure
            AllChem.UFFOptimizeMolecule(mol)

            # Compute Gasteiger partial charges
            AllChem.ComputeGasteigerCharges(mol)

            # Write the molecule to a PDB file
            with open(pdb_file, 'w') as f:
                f.write(Chem.MolToPDBBlock(mol))

            # Convert PDB to PDBQT using Pybel (Open Babel)
            ob_mol = next(pybel.readfile("pdb", pdb_file))
            ob_mol.write("pdbqt", pdbqt_file, overwrite=True)

            print(f"Successfully processed {compound_name}. PDBQT file saved as '{pdbqt_file}'.")

        except Exception as e:
            print(f"Error processing {compound_name}: {e}")
```

```python
from rdkit import Chem
from rdkit.Chem import AllChem
import subprocess
import os
from IPython.display import FileLink

def smiles_to_pdbqt(smiles, mol_name, output_dir="./output"):
    os.makedirs(output_dir, exist_ok=True)
    
    # Step 1: Generate RDKit mol with 3D coordinates
    mol = Chem.MolFromSmiles(smiles)
    mol = Chem.AddHs(mol)
    if AllChem.EmbedMolecule(mol) != 0:
        raise ValueError("3D embedding failed.")
    AllChem.UFFOptimizeMolecule(mol)

    # Step 2: Save as .mol file
    mol_path = os.path.join(output_dir, f"{mol_name}.mol")
    Chem.MolToMolFile(mol, mol_path)

    # Step 3: Convert to .pdbqt using Open Babel
    pdbqt_path = os.path.join(output_dir, f"{mol_name}.pdbqt")
    subprocess.run(["obabel", mol_path, "-O", pdbqt_path], check=True)

    # Step 4: Create download link
    return FileLink(pdbqt_path)

# Example usage
smiles = "CC1=C(C(=O)C(CO1)O)O"
mol_name = "dihydroxy_methyl_pyranone"
smiles_to_pdbqt(smiles, mol_name)
```
